## CÂU 3 (Yêu cầu c): Bóc tách dữ liệu cột time thành thời điểm trong ngày tính theo phút (minutes)
- Từ cột `time`, bóc tách thành số phút trôi qua từ đầu ngày (0h00):
  $$\text{minutes} = \text{hour} \times 60 + \text{minute} + \frac{\text{second}}{60}$$
- Đặt tên cột mới là `minutes`


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path('tb_tracking.xlsx')
df = pd.read_excel(path, header=None)
df.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
df['time'] = pd.to_datetime(df['time'])
df_filtered = df[df['time'] >= '2018-01-03'].copy()

# Bóc tách thời điểm đo đạc trong ngày tính theo phút
df_filtered['minutes'] = (
    df_filtered['time'].dt.hour * 60 + 
    df_filtered['time'].dt.minute + 
    df_filtered['time'].dt.second / 60.0
)

print("--- THỐNG KÊ MÔ TẢ CỘT MINUTES MỚI TẠO ---")
print(df_filtered['minutes'].describe())

print("\n--- 10 DÒNG ĐẦU KÈM THEO CỘT MINUTES ---")
print(df_filtered[['id', 'time', 'minutes', 'co_level', 'humidity', 'temperature']].head(10))
```

### Nhận xét Câu 3:
- Cột `minutes` chuyển đổi mốc thời gian ngày-tháng-năm thành một trục số liên tục trong ngày (khoảng giá trị từ 0 đến 1.440 phút).
- Trong tập dữ liệu đã lọc, các mốc đo nằm trong khoảng từ phút thứ 791.12 (khoảng 13h11 chiều) đến 1.439.40 (khoảng 23h59 đêm), rất thuận tiện để vẽ đồ thị theo dõi sự thay đổi theo chu kỳ ca làm việc trong ngày.
