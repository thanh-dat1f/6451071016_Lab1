- Đọc tập tin dữ liệu đính kèm `tb_tracking.xlsx`
- Thêm vào tiêu đề các cột theo thứ tự:
  1. `id`: định danh của dòng dữ liệu
  2. `device_id`: định danh của thiết bị đo
  3. `co_level`: nồng độ khí CO
  4. `humidity`: độ ẩm không khí
  5. `temperature`: nhiệt độ
  6. `time`: thời điểm đo đạc


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path('tb_tracking.xlsx')
# Đọc file không lấy dòng đầu làm header để tránh mất dữ liệu
df = pd.read_excel(path, header=None)

# Đặt lại tên cột theo yêu cầu
df.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']

print("--- THÔNG TIN CẤU TRÚC DỮ LIỆU SAU KHI GÁN TIÊU ĐỀ ---")
print(f"Số dòng: {df.shape[0]}, Số cột: {df.shape[1]}")
print("\n--- 10 DÒNG ĐẦU TIÊN CỦA BẢNG DỮ LIỆU ---")
print(df.head(10))

print("\n--- THÔNG TIN KIỂU DỮ LIỆU (df.info()) ---")
df.info()
```
