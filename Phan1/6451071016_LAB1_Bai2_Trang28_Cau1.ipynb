- Đọc tập tin dữ liệu `bank.csv` (sử dụng dấu phân cách `;`)
- Kiểm tra kích thước dữ liệu (shape), các dòng đầu tiên (`head()`)
- Xem cấu trúc và kiểu dữ liệu (`info()`)

```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path(os.path.join('bank', 'bank.csv'))
if not os.path.exists(path):
    path = get_data_path('bank.csv')

df = pd.read_csv(path, sep=';')

print("--- KÍCH THƯỚC BỘ DỮ LIỆU ---")
print(f"Số dòng: {df.shape[0]}, Số cột: {df.shape[1]}")

print("\n--- 10 DÒNG ĐẦU CỦA BỘ DỮ LIỆU ---")
print(df.head(10))

print("\n--- THÔNG TIN CẤU TRÚC VÀ KIỂU DỮ LIỆU (df.info()) ---")
df.info()

print("\n--- THỐNG KÊ MÔ TẢ THUỘC TÍNH ĐỊNH LƯỢNG (df.describe()) ---")
print(df.describe())
```
