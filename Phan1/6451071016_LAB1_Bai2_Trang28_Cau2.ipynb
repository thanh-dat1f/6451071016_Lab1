## CÂU 2: Thống kê dữ liệu thiếu và giá trị 'unknown'
- Trong tập dữ liệu Bank Marketing, dữ liệu thiếu được mã hóa dưới dạng chuỗi `'unknown'` thay vì giá trị `NaN` rỗng chuẩn.
- Thống kê số lượng và tỷ lệ % giá trị `'unknown'` trên từng thuộc tính.


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path(os.path.join('bank', 'bank.csv'))
if not os.path.exists(path):
    path = get_data_path('bank.csv')

df = pd.read_csv(path, sep=';')

unknown_counts = {}
unknown_percent = {}
n_rows = len(df)

for col in df.columns:
    if df[col].dtype == 'object':
        cnt = (df[col] == 'unknown').sum()
        unknown_counts[col] = cnt
        unknown_percent[col] = round((cnt / n_rows) * 100, 2)
    else:
        unknown_counts[col] = 0
        unknown_percent[col] = 0.0

unknown_df = pd.DataFrame({
    'Unknown_Count': unknown_counts,
    '%_Unknown': unknown_percent
}).sort_values(by='Unknown_Count', ascending=False)

print("--- THỐNG KÊ CÁC THUỘC TÍNH CÓ CHỨA GIÁ TRỊ 'unknown' ---")
print(unknown_df[unknown_df['Unknown_Count'] > 0])
```

### Nhận xét Câu 2:
- Thuộc tính `poutcome` (kết quả chiến dịch trước) có số lượng `unknown` áp đảo: 3.705 giá trị (chiếm 81.95%), do đa số khách hàng là người mới chưa từng tiếp cận trước đó.
- Thuộc tính `contact` (phương thức liên lạc) có 1.324 giá trị `unknown` (chiếm 29.29%).
- Thuộc tính `education` có 187 giá trị `unknown` (chiếm 4.14%).
- Thuộc tính `job` có 38 giá trị `unknown` (chiếm 0.84%).
=> Với `job` và `education`, tỷ lệ `unknown` nhỏ có thể thay thế bằng giá trị phổ biến nhất (mode). Với `poutcome` và `contact`, `unknown` mang ý nghĩa là một phân loại riêng biệt.
