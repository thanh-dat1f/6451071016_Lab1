## CÂU 5: Tiền xử lý dữ liệu Bank Marketing
- Thay thế giá trị `'unknown'` cho các cột có tỷ lệ nhỏ (`job`, `education`) bằng giá trị xuất hiện nhiều nhất (mode)
- Giới hạn ngoại lai (capping outliers) cho thuộc tính số dư (`balance`) và thời lượng cuộc gọi (`duration`)
- Kiểm tra lại tập dữ liệu sau xử lý


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path(os.path.join('bank', 'bank.csv'))
if not os.path.exists(path):
    path = get_data_path('bank.csv')

df = pd.read_csv(path, sep=';')

# 1. Điền giá trị phổ biến nhất (mode) cho job và education khi gặp 'unknown'
mode_job = df[df['job'] != 'unknown']['job'].mode()[0]
mode_edu = df[df['education'] != 'unknown']['education'].mode()[0]

df['job'] = df['job'].replace('unknown', mode_job)
df['education'] = df['education'].replace('unknown', mode_edu)

# 2. Xử lý ngoại lai (Capping outliers) theo phân vị 1% và 99% cho balance và duration
q_low = df['balance'].quantile(0.01)
q_high = df['balance'].quantile(0.99)
df['balance'] = df['balance'].clip(lower=q_low, upper=q_high)

q_dur_high = df['duration'].quantile(0.99)
df['duration'] = df['duration'].clip(upper=q_dur_high)

print(f"Thay thế 'unknown' ở job bằng: '{mode_job}'")
print(f"Thay thế 'unknown' ở education bằng: '{mode_edu}'")
print("--- THỐNG KÊ SAU KHI XỬ LÝ NGOẠI LAI VÀ UNKNOWN ---")
print(df[['balance', 'duration']].describe())
```

### Nhận xét Câu 5:
- Thuộc tính `job` và `education` đã được làm sạch, không còn chứa giá trị thiếu `'unknown'`.
- Thuộc tính `balance` và `duration` đã được xử lý cắt ngưỡng ngoại lai (capping at 1st and 99th percentiles) giúp dữ liệu ổn định, hạn chế các giá trị quá dị biệt gây sai lệch mô hình.
