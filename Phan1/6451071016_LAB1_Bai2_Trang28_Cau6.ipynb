## CÂU 6: Biến đổi thuộc tính (Feature Transformation & Scaling)
- Mã hóa biến mục tiêu `y` thành nhị phân (0: no, 1: yes)
- Mã hóa các biến nhị phân: `default`, `housing`, `loan` thành 0/1
- Phân nhóm (Binning) độ tuổi `age` thành 4 nhóm
- Chuẩn hóa Min-Max normalization cho các biến định lượng: `balance`, `duration`
- Áp dụng Label Encoding / One-hot encoding cho các biến phân loại còn lại


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path(os.path.join('bank', 'bank.csv'))
if not os.path.exists(path):
    path = get_data_path('bank.csv')

df = pd.read_csv(path, sep=';')

# 1. Mã hóa nhị phân biến mục tiêu y và các biến vay nợ
df['y'] = (df['y'] == 'yes').astype(int)
df['default'] = (df['default'] == 'yes').astype(int)
df['housing'] = (df['housing'] == 'yes').astype(int)
df['loan'] = (df['loan'] == 'yes').astype(int)

# 2. Phân nhóm độ tuổi (Age Binning)
# Nhóm 0: < 30 tuổi (Thanh niên)
# Nhóm 1: 30 - 45 tuổi (Trưởng thành)
# Nhóm 2: 45 - 60 tuổi (Trung niên)
# Nhóm 3: > 60 tuổi (Cao tuổi)
df['age_group'] = pd.cut(df['age'], bins=[0, 30, 45, 60, 100], labels=[0, 1, 2, 3]).astype(int)

# 3. Chuẩn hóa Min-Max normalization: new_val = (val - min) / (max - min)
for col in ['balance', 'duration']:
    col_min = df[col].min()
    col_max = df[col].max()
    df[col + '_normalized'] = (df[col] - col_min) / (col_max - col_min)

# 4. One-hot encoding cho các biến phân loại đa lớp
categorical_cols = ['job', 'marital', 'education', 'contact', 'poutcome']
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True, dtype=int)

print("--- KÍCH THƯỚC BỘ DỮ LIỆU SAU BIẾN ĐỔI ---")
print(f"Số dòng: {df_encoded.shape[0]}, Số cột: {df_encoded.shape[1]}")

print("\n--- 5 DÒNG ĐẦU TIÊN CỦA BỘ DỮ LIỆU ĐÃ BIẾN ĐỔI HOÀN CHỈNH ---")
print(df_encoded[['age', 'age_group', 'housing', 'loan', 'balance_normalized', 'duration_normalized', 'y']].head(5))
```

### Nhận xét Câu 6:
- Tập dữ liệu đã được số hóa hoàn toàn (all numerical features), sẵn sàng cung cấp trực tiếp cho các mô hình phân loại (Classification) hoặc phân cụm (Clustering).
- Việc chia nhóm độ tuổi và chuẩn hóa khoảng `[0, 1]` cho `balance`, `duration` giúp loại bỏ hoàn toàn sự chênh lệch đơn vị đo lường giữa các biến.
