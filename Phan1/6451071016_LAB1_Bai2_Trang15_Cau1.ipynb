## CÂU 1: Nhập dữ liệu và xem xét thông tin cơ bản của các thuộc tính
- Đọc tập dữ liệu `titanic.csv`
- Kiểm tra các dòng đầu tiên bằng hàm `head()`
- Xem thông tin tổng quan cấu trúc dữ liệu bằng hàm `info()`
- Thống kê mô tả các thuộc tính định lượng bằng hàm `describe()`


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


# 1. Đọc dữ liệu đầu vào
path = get_data_path('titanic.csv')
with open(path) as f:
    df = pd.read_csv(f, dtype={'age': np.float64, 'sibsp': np.int64, 'parch': np.int64})

print("--- 10 DÒNG ĐẦU TIÊN CỦA TẬP DỮ LIỆU ---")
print(df.head(10))

print("\n--- CẤU TRÚC VÀ KIỂU DỮ LIỆU (df.info()) ---")
df.info()

print("\n--- THỐNG KÊ MÔ TẢ CÁC THUỘC TÍNH ĐỊNH LƯỢNG (df.describe()) ---")
print(df.describe())
```
