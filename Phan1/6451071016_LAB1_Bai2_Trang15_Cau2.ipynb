## CÂU 2: Liệt kê danh sách 5 thuộc tính có số lượng giá trị rỗng nhiều nhất
- Đếm số lượng giá trị rỗng của mỗi cột
- Tính tỷ lệ phần trăm (%) giá trị rỗng
- Lấy ra top 5 thuộc tính rỗng nhiều nhất và đưa ra nhận xét


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path('titanic.csv')
with open(path) as f:
    df = pd.read_csv(f, dtype={'age': np.float64, 'sibsp': np.int64, 'parch': np.int64})

total = df.isnull().sum().sort_values(ascending=False)
percent_1 = df.isnull().sum() / df.isnull().count() * 100
percent_2 = (round(percent_1, 1)).sort_values(ascending=False)
missing_data = pd.concat([total, percent_2], axis=1, keys=['Total', '%'])

print("--- TOP 5 THUỘC TÍNH CÓ SỐ LƯỢNG GIÁ TRỊ RỖNG NHIỀU NHẤT ---")
print(missing_data.head(5))
```

- Thuộc tính `cabin` thiếu 1.014 giá trị rỗng (77.5%).
- Thuộc tính `boat` thiếu 823 giá trị rỗng (62.9%).
- Thuộc tính `home.dest` thiếu 564 giá trị rỗng (43.1%).

=> Các thuộc tính như `body`, `boat`, `home.dest` thiếu quá nhiều nên có thể loại bỏ; thuộc tính `cabin` có thể bóc tách chữ cái đầu làm boong tàu (`deck`); thuộc tính `age` cần được điền giá trị thiếu (imputation).
