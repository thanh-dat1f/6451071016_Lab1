## CÂU 2 (Yêu cầu b): Lọc bỏ dữ liệu không chính xác trước ngày 03/01/2018
- Do điều chỉnh cảm biến nên dữ liệu đo được bắt đầu từ ngày 03/01/2018 mới chính xác.
- Chuyển đổi cột `time` sang kiểu datetime.
- Loại bỏ toàn bộ các dòng đo đạc trước ngày 03/01/2018.


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path('tb_tracking.xlsx')
df = pd.read_excel(path, header=None)
df.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']

# 1. Chuyển đổi cột time sang datetime
df['time'] = pd.to_datetime(df['time'])

# 2. Lọc bỏ các dòng dữ liệu trước ngày 03/01/2018
initial_count = len(df)
df_filtered = df[df['time'] >= '2018-01-03'].copy()
filtered_count = len(df_filtered)
removed_count = initial_count - filtered_count

print("--- THỐNG KÊ QUÁ TRÌNH LỌC DỮ LIỆU ---")
print(f"• Số lượng bản ghi ban đầu: {initial_count} dòng")
print(f"• Số lượng bản ghi hợp lệ từ ngày 03/01/2018: {filtered_count} dòng")
print(f"• Số lượng bản ghi không chính xác đã loại bỏ: {removed_count} dòng ({round(removed_count/initial_count*100, 2)}%)")

print("\n--- 5 DÒNG ĐẦU DỮ LIỆU HỢP LỆ BẮT ĐẦU TỪ 03/01/2018 ---")
print(df_filtered.head(5))

print("\n--- THỜI GIAN ĐO SỚM NHẤT VÀ MUỘN NHẤT TRONG DỮ LIỆU ĐÃ LỌC ---")
print(f"Thời điểm bắt đầu: {df_filtered['time'].min()}")
print(f"Thời điểm kết thúc: {df_filtered['time'].max()}")
```

### Nhận xét Câu 2:
- Toàn bộ dữ liệu trước ngày 03/01/2018 (31.422 dòng, chiếm 94.25%) đã được loại bỏ thành công.
- Dữ liệu hợp lệ còn lại là 1.918 dòng được ghi nhận liên tục từ 13:11:07 ngày 03/01/2018, đảm bảo tính chuẩn xác cho việc phân tích môi trường nhà xưởng.
