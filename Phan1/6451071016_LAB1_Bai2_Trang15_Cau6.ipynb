## CÂU 6: Tiền xử lý các thuộc tính thiếu dữ liệu và xóa bỏ thuộc tính không cần thiết
- Trích xuất boong tàu `deck` từ ký tự đầu của `cabin` trước khi xóa `cabin`
- Xóa các thuộc tính không cần thiết: `ticket`, `boat`, `body`, `home.dest`
- Điền giá trị thiếu vào thuộc tính tuổi (`age`) dựa trên mean và std
- Điền giá trị thiếu vào Cổng lên tàu (`embarked`) bằng giá trị phổ biến nhất `'S'`
- Điền giá trị thiếu cho `fare` bằng 0 / median


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename

import re

path = get_data_path('titanic.csv')
with open(path) as f:
    df = pd.read_csv(f, dtype={'age': np.float64, 'sibsp': np.int64, 'parch': np.int64})

# 1. Trích xuất thuộc tính deck từ cabin
deck = {"A": 1, "B": 2, "C": 3, "D": 4, "E": 5, "F": 6, "G": 7, "U": 8}
for dataset in [df]:
    dataset['cabin'] = dataset['cabin'].fillna("U0")
    dataset['deck'] = dataset['cabin'].map(lambda x: re.compile("([a-zA-Z]+)").search(x).group())
    dataset['deck'] = dataset['deck'].map(deck)
    dataset['deck'] = dataset['deck'].fillna(0)
    dataset['deck'] = dataset['deck'].astype(int)

# 2. Xóa các cột không cần thiết
df = df.drop(['cabin'], axis=1)
df = df.drop(['ticket'], axis=1)
df = df.drop(['boat'], axis=1)
df = df.drop(['body'], axis=1)
df = df.drop(['home.dest'], axis=1)

# 3. Điền giá trị còn thiếu cho age
np.random.seed(42)
for dataset in [df]:
    mean = df["age"].mean()
    std = df["age"].std()
    is_null = dataset["age"].isnull().sum()
    rand_age = np.random.randint(int(mean - std), int(mean + std), size=is_null)
    age_slice = dataset["age"].copy()
    age_slice[np.isnan(age_slice)] = rand_age
    dataset["age"] = age_slice
    dataset["age"] = df["age"].astype(int)

# 4. Điền giá trị còn thiếu cho embarked bằng giá trị phổ biến nhất 'S'
common_value = 'S'
for dataset in [df]:
    dataset['embarked'] = dataset['embarked'].fillna(common_value)

# 5. Điền giá trị còn thiếu cho fare
for dataset in [df]:
    dataset['fare'] = dataset['fare'].fillna(0)

print("--- SỐ LƯỢNG GIÁ TRỊ RỖNG CỦA CÁC THUỘC TÍNH SAU TIỀN XỬ LÝ ---")
print(df.isnull().sum())

print("\n--- CẤU TRÚC TẬP DỮ LIỆU HIỆN TẠI (df.info()) ---")
df.info()
```

### Nhận xét Câu 6:
- Các cột rỗng nhiều và không giúp ích cho phân loại như `cabin`, `ticket`, `boat`, `body`, `home.dest` đã được loại bỏ an toàn.
- Cột `deck` được tạo ra bảo toàn thông tin boong tàu của hành khách.
- Các cột quan trọng `age`, `embarked`, `fare` đã được điền đủ dữ liệu, tập dữ liệu không còn giá trị rỗng (NaN).
