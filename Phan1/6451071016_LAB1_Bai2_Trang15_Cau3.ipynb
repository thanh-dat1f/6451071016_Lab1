## CÂU 3: Khảo sát sự liên quan giữa Độ tuổi (age), Giới tính (sex) và Khả năng sống sót (survived)
- Vẽ biểu đồ phân bố độ tuổi sống sót và tử vong riêng biệt cho Nam và Nữ
- Đưa ra nhận xét chi tiết


```python
# ==========================================
# THÔNG TIN SINH VIÊN THỰC HIỆN
# Họ và tên: Hoàng Thành Đạt
# MSSV: 6451071016
# ==========================================
ho_ten = "Hoàng Thành Đạt"
mssv = "6451071016"
print(f"Họ tên: {ho_ten} - MSSV: {mssv}")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def get_data_path(rel_filename):
    for candidate in [
        os.path.join("..", "..", "dataset", rel_filename),
        os.path.join("..", "dataset", rel_filename),
        os.path.join("dataset", rel_filename),
        rel_filename
    ]:
        if os.path.exists(candidate):
            return candidate
    return rel_filename


path = get_data_path('titanic.csv')
with open(path) as f:
    df = pd.read_csv(f, dtype={'age': np.float64, 'sibsp': np.int64, 'parch': np.int64})

survived = 'survived'
not_survived = 'not survived'
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))
women = df[df['sex'] == 'female']
men = df[df['sex'] == 'male']

ax = sns.histplot(women[women['survived'] == 1].age.dropna(), bins=18, label=survived, ax=axes[0], kde=False, color='g')
ax = sns.histplot(women[women['survived'] == 0].age.dropna(), bins=40, label=not_survived, ax=axes[0], kde=False, color='y')
ax.legend()
ax.set_title('Female')

ax = sns.histplot(men[men['survived'] == 1].age.dropna(), bins=18, label=survived, ax=axes[1], kde=False, color='g')
ax = sns.histplot(men[men['survived'] == 0].age.dropna(), bins=40, label=not_survived, ax=axes[1], kde=False, color='y')
ax.legend()
_ = ax.set_title('Male')
plt.tight_layout()
plt.show()
```

### Nhận xét Câu 3:
- Xác suất nam sống sót cao nhất ở lứa tuổi từ 1 đến 10.
- Cơ hội sống sót của nữ cao hơn rõ rệt trong khoảng từ 12 đến 40 tuổi.
- Xác suất nam không sống sót cao nhất ở độ tuổi thanh niên từ 18 đến 27 tuổi.
- Nhìn chung, tỷ lệ sống sót của nữ ở hầu hết mọi lứa tuổi đều vượt trội so với nam (phù hợp với quy tắc đạo đức hàng hải ưu tiên cứu phụ nữ và trẻ em).
